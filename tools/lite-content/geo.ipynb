# Geodata on the canvas

GeoPandas runs in this tab, and `trazo.map` draws a GeoDataFrame as a hand-drawn map in Trazo's canvas (Python → canvas). There is no basemap: only your geometry, so nothing is fetched from anywhere.

````python
import geopandas as gpd
import numpy as np
from shapely.geometry import box, LineString, Point
import trazo

rng = np.random.default_rng(3)
names = ['North', 'Harbour', 'Old town', 'Park', 'Market', 'Station', 'Hills', 'Lakeside', 'East', 'South', 'Docks', 'West']
cells = [box(-77.10 + i * 0.045, -12.14 + j * 0.04, -77.10 + (i + 1) * 0.045, -12.14 + (j + 1) * 0.04) for j in range(3) for i in range(4)]
districts = gpd.GeoDataFrame({'name': names, 'density': rng.integers(2000, 21000, 12)}, geometry=cells, crs=4326)
districts.head()
````

````python
river = gpd.GeoSeries([LineString([(-77.10, -12.12), (-77.04, -12.09), (-76.99, -12.1), (-76.93, -12.04)])], crs=4326)
stations = gpd.GeoSeries([Point(-77.08, -12.12), Point(-77.03, -12.05), Point(-76.98, -12.09)], crs=4326)
````

````python
# a list of layers is drawn together; `color` paints by a column (five classes for numbers), `label` writes a column on each shape
trazo.map([districts, river, stations], color='density', label='name', title='Districts by density')
````

Reproject first when you want projected coordinates (they are drawn as they are, in metres):

```python
trazo.map(districts.to_crs(32718), color='density', title='UTM 18S')
```
