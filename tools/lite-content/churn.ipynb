# Customer churn (example)

Try **Diagram this notebook** in Trazo's panel: it draws the data flow of this notebook (tables, merge, model, predictions).

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(0)
customers = pd.DataFrame({'id': range(200), 'plan': rng.choice(['free', 'pro'], 200)})
orders = pd.DataFrame({'id': rng.integers(0, 200, 400), 'price': rng.uniform(5, 90, 400), 'qty': rng.integers(1, 5, 400)})
```

```python
df = customers.merge(orders, on='id', how='left')
df = df.dropna()
df['total'] = df['price'] * df['qty']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df[['total', 'qty']]
y = df['plan']
X_train, X_test, y_train, y_test = train_test_split(X, y)
model = RandomForestClassifier()
model.fit(X_train, y_train)
pred = model.predict(X_test)
print((pred == y_test).mean())
```
