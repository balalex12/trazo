# Python in your browser

This notebook runs **in this tab**: Python, pandas and matplotlib come from your own machine and nothing is sent anywhere. Run a cell with Shift+Enter.

In Trazo's panel, **Diagram this notebook** draws where your data goes, and **Send diagram to notebook** adds your canvas here as pandas tables.

```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(1)
sales = pd.DataFrame({
    'region': rng.choice(['north', 'south', 'east'], 60),
    'units': rng.integers(1, 20, 60),
    'price': rng.uniform(5, 50, 60).round(2),
})
sales['total'] = sales['units'] * sales['price']
sales.head()
```

```python
by_region = sales.groupby('region').agg({'total': 'sum'})
by_region
```

```python
import matplotlib.pyplot as plt

by_region.plot.bar(legend=False)
plt.ylabel('total')
plt.show()
```

## Talk to the canvas

Inside Trazo's panel, Python can draw on the canvas and read it back:

```python
import trazo

trazo.draw({
    "title": "Sales regions",
    "nodes": [{"id": r, "label": r} for r in by_region.index] + [{"id": "all", "label": "all regions", "role": "database"}],
    "edges": [(r, "all") for r in by_region.index],
})
```

```python
nodes, edges = trazo.canvas()   # the diagram on the canvas, as of the end of the previous cell
nodes
```
