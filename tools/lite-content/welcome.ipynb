```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(1)
sales = pd.DataFrame({
    'region': rng.choice(['north', 'south', 'east'], 60),
    'units': rng.integers(1, 20, 60),
    'price': rng.uniform(5, 50, 60).round(2),
})
sales['total'] = sales['units'] * sales['price']
sales.head()
```

```python
by_region = sales.groupby('region').agg({'total': 'sum'})
by_region
```

```python
import matplotlib.pyplot as plt

by_region.plot.bar(legend=False)
plt.ylabel('total')
plt.show()
```
